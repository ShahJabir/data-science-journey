```python
"""Pandas Tutorial"""
import numpy as np
import pandas as pd
```

## Creating Series
There are multiple ways to create a pandas Series

```python
labels = ['a','b','c']
my_list = [10, 20, 30]
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr, index=labels)
```

```python
pd.Series(d)
```

## Data Frame

### Creaing a DattaFrame

```python
data = {
  'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
  'Age': [25, 30, 35, 40, 45],
  'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
  'Salary': [50000, 60000, 70000, 80000, 90000]
}
df = pd.DataFrame(data)
print(df)
```

```python
data_list = [
  ['Alice', 25, 'New York', 50000],
  ['Bob', 30, 'Los Angeles', 60000],
  ['Charlie', 35, 'Chicago', 70000],
  ['David', 40, 'Houston', 80000],
  ['Eve', 45, 'Phoenix', 90000],
]
df_l = pd.DataFrame(data_list, columns=['Name', 'Age', 'City', 'Salary'])
print(df_l)
```
